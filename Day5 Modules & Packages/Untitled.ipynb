```python
# Module
# Module = ek Python file (.py) jisme code/functions/classes hote hain.
# ex...
# math.py
# __________________________________________________
# Is file ke andar kuch functions ho sakte hain:
# __________________________________________________
# def add(a, b):
#     return a + b
# __________________________________________________

# Import
# Doosri Python file/module ka code apne program mein use karna.
# __________________________________________________
# import math
# print(math.sqrt(25))
# __________________________________________________
# math → module
# sqrt() → us module ka function

# Built-in Modules
# Python ke saath already bahut saare modules milte hain.
# __________________________________________________
import math
print(math.sqrt(16))
print(math.pi)
print(math.pow(2, 3)) #matlab hai 2 ki power 3 
# __________________________________________________

# Random
# Random number generate karne ke liye:
# __________________________________________________
import random
print(random.randint(1, 10))
print(random.randint(1, 100))
# __________________________________________________

# DateTime
# Date aur time ke liye hota hai.
# __________________________________________________
import datetime    #Ye current date/time dega.
today = datetime.datetime.now()
print("Today:",today)
# __________________________________________________


```
