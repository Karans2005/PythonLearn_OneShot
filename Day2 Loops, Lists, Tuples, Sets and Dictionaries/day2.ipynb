```python
# Loops

for i in range(5):
     print('number',i)
for i in range(1,11):
     print(i)

# Loops with List

names = ['SRK','AMIT','RAJESH']
for name in names:
     print(names)
for name in names:
     print(name)
    

# While Loops

i = 1
while i <= 5:
    print(i)
    i = i + 1

# LIST IN PYTHON

fruits = ["Apple", "Banana", "Mango"]

print(fruits)
print(fruits[0])   #for index
print(fruits[1])
print(fruits[2])

# APPEND METHOD

fruits.append("Orange")

print(fruits)

# REMOVE METHOD

# fruits.remove("Orange")
# fruits.remove("Banana")
# print(fruits)

# Find length
print(len(fruits))

fruits = ["Apple", "Banana", "Mango"]

for fruit in fruits:
    print(fruit)

# TUPPLE METHOD
# Tuple bhi multiple values store karta hai, lekin normally change nahi kar sakte.

numbers = (10, 20, 30, 40)

print(numbers)
print(numbers[0])

# List vs Tuple

# List → [] → values change kar sakte ho
# Tuple → () → values change nahi kar sakte

# List
# a = [10, 20, 30]
# a[0] = 100       # Allowed

# # Tuple
# b = (10, 20, 30)
# b[0] = 100       # Error

# SETS
# Set mein unique values store hoti hain.Duplicate values automatically remove ho jaati hain.

numbers = {10, 20, 30, 10, 20}
print(numbers)

numbers.add(40)
print(numbers)
numbers.remove(20)
print(numbers)

# DICTIONARIES ***
# Data ko key : value format mein store karta hai.

student = {
    "name": "Harsh",
    "age": 21,
    "city": "Durg"
}

print(student)
print(student["name"])
print(student["age"])

student["age"] = 22
student["city"] = "Bhilai"
print(student)

# Dictionary in loop
for key,value in student.items():
    print(key,value)

# NOTE:::
# Ek baar difference samjho
# Data Structure	Example	Main Feature
# List-	[]	Changeable
# Tuple- ()	Generally unchangeable
# Set-	{}	Unique values
# Dictionary- {key: value}	Key-value data
```
