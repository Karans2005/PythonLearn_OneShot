```python
# OOP (Object-Oriented Programming) ke ye concepts karenge:

# Class  # Object  # __init__  # self  # Methods  # Inheritance


# Class = blueprint / design
# Jaise ghar banane se pehle uska design hota hai, waise hi Object banane ke liye Class hoti hai.
# _____________________________________________
# class Student:   #Yahan Student ek Class hai.
#     pass
# _____________________________________________

# Object
# Class se banne wali actual cheez ko Object kehte hain.
# _____________________________________________
# class Student:
#     pass

# student1 = Student()
# student2 = Student()
# _____________________________________________
# Student → Class
# student1,student2 → Object
# Ek class se multiple objects bana sakte ho.

# __init__
# iska use object banate time initial data set karne ke liye hota hai.
# _____________________________________________
# class Student:

#     def __init__(self, name, age):
#         self.name = name
#         self.age = age
# _____________________________________________

# OBJECT:
# _____________________________________________
# class Student:

#     def __init__(self, name, age):
#         self.name = name
#         self.age = age
# student1 = Student("Harsh", 21)

# print(student1.name)
# print(student1.age)
# _____________________________________________

# Self
# self ka matlab hai current object.Is current object ka name, jo name value mili hai usko store karo.

# self.name = name
# To
# student1.name → Harsh
# student2.name → Rahul

# Methods
# Class ke andar banaya hua function Method kehlata hai.
# _____________________________________________
class Student:

    def __init__(self, name):
        self.name = name

    def greet(self):
        print("Hello", self.name)

student1 = Student("Harsh")

student1.greet()
# _____________________________________________

# Inheritance ⭐
# Ek class ki properties/methods ko doosri class mein use karna = Inheritance.
# _____________________________________________
class Animal:

    def speak(self):
        print("Animal makes sound")


class Dog(Animal):
    pass

dog = Dog()
dog.speak()    
# _____________________________________________

# OOP ka short summary
# Class       → Blueprint
# Object      → Class se bani actual cheez
# __init__    → Object ka initial data
# self        → Current object
# Method      → Class ke andar function
# Inheritance → Parent ka code Child mein use karna
```
