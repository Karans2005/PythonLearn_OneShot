```python
# Data Types

name = "Harsh"       # string
age = 21             # integer
price = 99.50        # float
is_student = True    # boolean

print(type(name))
print(type(age))
print(type(price))
print(type(is_student))
```

```python
# Input

name = input('Enter Your Name: ')
print('Name: ',name)

# Number input

age = int(input('Enter Your Age: '))
print('Age:',age)
```

```python
# Operators

a = 10
b = 3
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a%b)
print(a**b)
```
